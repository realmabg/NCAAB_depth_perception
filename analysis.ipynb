```python
import pandas as pd
```

```python
away = pd.read_csv("shots.csv")     
arenas = pd.read_csv("arenas.csv")     
```

```python
merged = away.merge(
    arenas.add_prefix("arena_"),
    left_on="homeMarket",
    right_on="arena_School",
    how="left",
    validate="m:1"
)

# (Optional) sanity: make sure all homes matched an arena
unmatched = merged["arena_School"].isna().sum()
print(f"Unmatched home teams: {unmatched}")
```
