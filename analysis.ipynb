```python
import numpy as np

# Does the shot’s basket have a wall?
merged["side_has_wall"] = np.where(
    merged["side"].eq("LEFT"),  merged["arena_Left_Wall"],
    np.where(merged["side"].eq("RIGHT"), merged["arena_Right_Wall"], np.nan)
)

# Distance for that basket (Close/Far/None)
merged["side_wall_distance"] = np.where(
    merged["side"].eq("LEFT"),  merged["arena_Left_Wall_Distance"],
    np.where(merged["side"].eq("RIGHT"), merged["arena_Right_Wall_Distance"], np.nan)
)
merged["side_wall_distance"] = np.where(
    merged["side_has_wall"] == True, merged["side_wall_distance"], "None"
)

# (Optional) Type for that basket (Whole/Mix/None)
merged["side_wall_type"] = np.where(
    merged["side"].eq("LEFT"),  merged["arena_Left_Wall_Type"],
    np.where(merged["side"].eq("RIGHT"), merged["arena_Right_Wall_Type"], np.nan)
)
merged["side_wall_type"] = np.where(
    merged["side_has_wall"] == True, merged["side_wall_type"], "None"
)
```

```python
zone_summary = (
    merged
    .groupby(
        ["homeMarket","season","zones13","side","similar_team",
         "side_wall_distance","side_wall_type"],
        dropna=False
    )
    .agg(attempts=("success","count"), makes=("success","sum"))
    .assign(FG_pct=lambda d: d["makes"]/d["attempts"])
    .reset_index()
)

similar_lookup = (
    zone_summary
    .groupby(["homeMarket","season","zones13","side"], as_index=False)
    .agg(sim_attempts=("attempts","sum"), sim_makes=("makes","sum"))
    .assign(FG_pct_similar=lambda d: d["sim_makes"]/d["sim_attempts"])
    .rename(columns={"homeMarket":"sim_team"})
    [["sim_team","season","zones13","side","FG_pct_similar","sim_attempts"]]
)

zone_with_similar = (
    zone_summary
    .merge(
        similar_lookup,
        left_on=["similar_team","season","zones13","side"],
        right_on=["sim_team","season","zones13","side"],
        how="left",
        validate="m:1"
    )
)
zone_with_similar["diff_FG_pct"] = zone_with_similar["FG_pct"] - zone_with_similar["FG_pct_similar"]



by_distance = (
    zone_with_similar
    .groupby(["zones13","side_wall_distance"], dropna=False)
    .apply(lambda g: pd.Series({
        "focal_attempts": g["attempts"].sum(),
        "focal_makes": g["makes"].sum(),
        "focal_FG_pct": g["makes"].sum() / g["attempts"].sum() if g["attempts"].sum() else np.nan,
        "sim_attempts": g["sim_attempts"].sum(),
        "sim_makes": (g["FG_pct_similar"] * g["sim_attempts"]).sum(),
        "sim_FG_pct": (g["FG_pct_similar"] * g["sim_attempts"]).sum() / g["sim_attempts"].sum() if g["sim_attempts"].sum() else np.nan,
        "w_mean_diff": (g["diff_FG_pct"] * g["attempts"]).sum() / g["attempts"].sum()
    }))
    .reset_index()
)

by_distance_no_zone = (
    zone_with_similar
    .groupby(["side_wall_distance"], dropna=False)
    .apply(lambda g: pd.Series({
        "focal_attempts": g["attempts"].sum(),
        "focal_makes": g["makes"].sum(),
        "focal_FG_pct": g["makes"].sum() / g["attempts"].sum() if g["attempts"].sum() else np.nan,
        "sim_attempts": g["sim_attempts"].sum(),
        "sim_makes": (g["FG_pct_similar"] * g["sim_attempts"]).sum(),
        "sim_FG_pct": (g["FG_pct_similar"] * g["sim_attempts"]).sum() / g["sim_attempts"].sum() if g["sim_attempts"].sum() else np.nan,
        "w_mean_diff": (g["diff_FG_pct"] * g["attempts"]).sum() / g["attempts"].sum()
    }))
    .reset_index()
)

by_type = (
    zone_with_similar
    .groupby(["zones13","side_wall_type"], dropna=False)
    .apply(lambda g: pd.Series({
        "focal_attempts": g["attempts"].sum(),
        "focal_makes": g["makes"].sum(),
        "focal_FG_pct": g["makes"].sum() / g["attempts"].sum() if g["attempts"].sum() else np.nan,
        "sim_attempts": g["sim_attempts"].sum(),
        "sim_makes": (g["FG_pct_similar"] * g["sim_attempts"]).sum(),
        "sim_FG_pct": (g["FG_pct_similar"] * g["sim_attempts"]).sum() / g["sim_attempts"].sum() if g["sim_attempts"].sum() else np.nan,
        "w_mean_diff": (g["diff_FG_pct"] * g["attempts"]).sum() / g["attempts"].sum()
    }))
    .reset_index()
)

by_type_no_zone = (
    zone_with_similar
    .groupby(["side_wall_type"], dropna=False)
    .apply(lambda g: pd.Series({
        "focal_attempts": g["attempts"].sum(),
        "focal_makes": g["makes"].sum(),
        "focal_FG_pct": g["makes"].sum() / g["attempts"].sum() if g["attempts"].sum() else np.nan,
        "sim_attempts": g["sim_attempts"].sum(),
        "sim_makes": (g["FG_pct_similar"] * g["sim_attempts"]).sum(),
        "sim_FG_pct": (g["FG_pct_similar"] * g["sim_attempts"]).sum() / g["sim_attempts"].sum() if g["sim_attempts"].sum() else np.nan,
        "w_mean_diff": (g["diff_FG_pct"] * g["attempts"]).sum() / g["attempts"].sum()
    }))
    .reset_index()
)
```

```python
def classify_group(g):
    # Rank distances in this zone (higher w_mean_diff = better)
    ranks = g["w_mean_diff"].rank(method="min", ascending=False)

    g = g.copy()
    g["category"] = "Middle"
    g.loc[ranks == 1, "category"] = "Best"
    g.loc[ranks == ranks.max(), "category"] = "Worst"
    return g

# Apply classification per zone
ranked_distance = by_distance.groupby("zones13", group_keys=False).apply(classify_group)

# Count across zones
summary_counts_distance = (
    ranked_distance.groupby(["side_wall_distance","category"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

ranked_type = by_type.groupby("zones13", group_keys=False).apply(classify_group)

# Count across zones
summary_counts_type = (
    ranked_type.groupby(["side_wall_type","category"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)
```

```python
by_type = (
    zone_with_similar
    .groupby(["zones13", "arena_Type of Wall"], dropna=False)
    .apply(lambda g: pd.Series({
        # focal totals
        "focal_attempts": g["attempts"].sum(),
        "focal_makes": g["makes"].sum(),
        "focal_FG_pct": g["makes"].sum() / g["attempts"].sum() if g["attempts"].sum() > 0 else np.nan,

        # similar totals (reconstruct makes)
        "sim_attempts": g["sim_attempts"].sum(),
        "sim_makes": (g["FG_pct_similar"] * g["sim_attempts"]).sum(),
        "sim_FG_pct": (g["FG_pct_similar"] * g["sim_attempts"]).sum() / g["sim_attempts"].sum() if g["sim_attempts"].sum() > 0 else np.nan,

        # weighted mean ΔFG% (weights = focal attempts)
        "w_mean_diff": (g["diff_FG_pct"] * g["attempts"]).sum() / g["attempts"].sum()
    }))
    .reset_index()
)

by_type_no_zone = (
    zone_with_similar
    .groupby(["arena_Type of Wall"], dropna=False)
    .apply(lambda g: (g["diff_FG_pct"] * g["attempts"]).sum() / g["attempts"].sum())
    .reset_index(name="w_mean_diff")
)
```

```python
zone_summary2 = (
    merged
    .groupby(["homeMarket","season","zones13","side","similar_team","side_wall_distance"], dropna=False)
    .agg(attempts=("success","count"), makes=("success","sum"))
    .assign(FG_pct=lambda d: d["makes"]/d["attempts"])
    .reset_index()
)
similar_lookup2 = (
    zone_summary2
    .groupby(["homeMarket","season","zones13","side"], as_index=False)
    .agg(sim_attempts=("attempts","sum"), sim_makes=("makes","sum"))
    .assign(FG_pct_similar=lambda d: d["sim_makes"]/d["sim_attempts"])
    .rename(columns={"homeMarket":"sim_team"})
    [["sim_team","season","zones13","side","FG_pct_similar","sim_attempts"]]
)
zone_with_similar2 = (
    zone_summary2
    .merge(
        similar_lookup,
        left_on=["similar_team","season","zones13","side"],
        right_on=["sim_team","season","zones13","side"],
        how="left",
        validate="m:1"
    )
)

zone_with_similar2["diff_FG_pct"] = zone_with_similar2["FG_pct"] - zone_with_similar2["FG_pct_similar"]

# Optional filters
zone_with_similar2 = zone_with_similar2.dropna(subset=["FG_pct_similar"])
zone_with_similar2 = zone_with_similar2.loc[
    (zone_with_similar2["attempts"] >= 10) & (zone_with_similar2["sim_attempts"] >= 10)
]

# Weighted mean ΔFG% by distance, per zone
by_distance = (
    zone_with_similar2
    .groupby(["zones13","side_wall_distance"], dropna=False)
    .apply(
        lambda g: pd.Series({
            # Total attempts and makes (from focal team)
            "total_attempts": g["attempts"].sum(),
            "total_makes": g["makes"].sum(),
            # Weighted mean ΔFG% (using attempts as weight)
            "w_mean_diff": (g["diff_FG_pct"] * g["attempts"]).sum() / g["attempts"].sum()
        })
    )
    .reset_index()
)
by_distance["focal_FG_pct"] = by_distance["total_makes"] / by_distance["total_attempts"]

by_distance_no_zone = (
    zone_with_similar2
    .groupby(["side_wall_distance"], dropna=False)
    .apply(
        lambda g: pd.Series({
            # Total attempts and makes (from focal team)
            "total_attempts": g["attempts"].sum(),
            "total_makes": g["makes"].sum(),
            # Weighted mean ΔFG% (using attempts as weight)
            "w_mean_diff": (g["diff_FG_pct"] * g["attempts"]).sum() / g["attempts"].sum()
        })
    )
    .reset_index()
)
by_distance_no_zone["focal_FG_pct"] = by_distance_no_zone["total_makes"] / by_distance_no_zone["total_attempts"]

```
