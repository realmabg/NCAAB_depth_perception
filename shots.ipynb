# Cleaning

```python
shots_df = pd.read_csv("pbp_shots.csv")
shots_df
```

```python
comp_to_season = {
    27693: 2021,
    30629: 2022,
    33533: 2023,
    36046: 2024,
    38409: 2025
}
shots_df["season"] = shots_df["competitionId"].map(comp_to_season)
```

```python
shots_df["homeMarket"].unique()
```

```python
jump_shots_list = ['jumpshot', 'stepbackjumpshot', 'hookshot',
       'floatingjumpshot', 'pullupjumpshot',
       'turnaroundjumpshot']
```

```python
jump_shots_df = shots_df[shots_df["subType"].isin(jump_shots_list)]
jump_shots_df
```

```python
teamdrtg = pd.read_csv("teamdrtg.csv")
teamdrtg
```

```python
pd.set_option('display.max_columns', None)
```

```python
drtg_pd = jump_shots_df.merge(
    teamdrtg,
    how="left",
    left_on=["homeMarket", "season"],
    right_on=["team", "Year"]
)
drtg_pd
```

```python
shots_less_info_df=drtg_pd.drop(columns=["competitionId","gameDate","homeId","awayId","actionNumber","teamId","playerId","fullName","x","y","xGs","yGs","adjde","similar_adjde","adjde_diff"])
```

```python
shots_less_info_df
```

```python
away_shots_df = shots_less_info_df[shots_less_info_df["homeMarket"]!=shots_less_info_df["teamMarket"]]
```

```python
def check_half_and_side(away_shots_df, team_col="teamMarket"):
    df = away_shots_df.copy()


    df["half"] = np.select(
        [df["secsIntoGame"] <= 1200, df["secsIntoGame"] <= 2400],
        [1, 2],
        default=3
    )


    half_max = (
        df.groupby(["gameId", "half"])["secsIntoGame"]
          .max()
          .unstack(fill_value=np.nan)
    )
    bad_time_first  = half_max[half_max.get(1, pd.Series(dtype=float)) > 1200]
    bad_time_second = half_max[half_max.get(2, pd.Series(dtype=float)) <= 1200]

    side_nunique = (
        df.groupby(["gameId", team_col, "half"])["side"]
          .nunique(dropna=True)
          .rename("num_sides")
    )
    viol_side = side_nunique[side_nunique > 1]


    def first_mode(s: pd.Series):
        s = s.dropna()
        if s.empty: return np.nan
        m = s.mode()
        return m.iloc[0] if not m.empty else np.nan

    primary_side = (
        df.groupby(["gameId", team_col, "half"])["side"]
          .agg(first_mode)
          .rename("primary_side")
          .reset_index()
    )

    df = df.merge(primary_side, on=["gameId", team_col, "half"], how="left")
    df["side_mismatch"] = (df["side"] != df["primary_side"]) & df["primary_side"].notna()


    mismatched_rows = df[df["side_mismatch"]].copy()


    clean_df = df[~(df.index.isin(mismatched_rows.index) & df["half"].isin([1, 2]))].copy()

    return {
        "half_max_per_game": half_max,
        "bad_time_first": bad_time_first,
        "bad_time_second": bad_time_second,
        "viol_side_counts": viol_side,
        "mismatched_rows": mismatched_rows,
        "df_with_primary": df,
        "clean_df": clean_df,                  
    }
```

```python
res = check_half_and_side(away_shots_df, team_col="teamMarket")
clean_away_shots_df = res["clean_df"]  # OT kept, bad H1/H2 shots dropped
```

```python
clean_away_shots_df
```

```python
clean_away_shots_df.to_csv("shots.csv",index=False)
```
